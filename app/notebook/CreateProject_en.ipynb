# ***Create Project***

### 1. Enter the analysis project name as in the example
- Example: projectdir = "new_analysis" <br>
        → A folder named "new_analysis" will be created
- Do not include spaces
- Do not use existing folder names

### 2. Click RUN (▶) at the top of this notebook

```python
### Modify "new_analysis" to any name ###
projectdir = "new_analysis"

### Do not modify the following ###
import shutil

project_dir = f'./project/{projectdir}'

try:
    shutil.copytree('./template/en', project_dir)
    print(f'Project folder "{project_dir}" has been successfully created.')
    print('\n\033[48;5;214m\033[30m\033[1m ****************************** Complete! ****************************** \033[0m')
except FileExistsError:
    print(f'Error: Project folder "{project_dir}" already exists. Please specify a different name.')
except Exception as e:
    print(f'Error: Failed to create project folder ({e})')
```
